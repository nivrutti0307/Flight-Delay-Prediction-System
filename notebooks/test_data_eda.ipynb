```python
# Fill missing delay causes with 0
delay_cols = ['CarrierDelay', 'WeatherDelay', 'NASDelay', 'SecurityDelay', 'LateAircraftDelay']
df[delay_cols] = df[delay_cols].fillna(0)

# Optional: Fill missing city/lat/long using mode (you can improve this by using a lookup later)
location_cols = ['Origin_City', 'Origin_Lat', 'Origin_Long', 'Dest_City', 'Dest_Lat', 'Dest_Long']
df[location_cols] = df[location_cols].fillna(method='ffill')  # quick fix for now

# Keep NaNs in ArrTime, AirTime, ArrDelay, etc. (they signal cancelled/diverted)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.histplot(df['DepDelay'].dropna(), bins=50)
plt.title("Distribution of Departure Delays (in minutes)")
plt.show()
```

```python
# Replace missing or invalid time codes (like 2400) with NaN
df['CRSDepTime'] = df['CRSDepTime'].apply(lambda x: x if x != 2400 else 0)
df['DepTime'] = df['DepTime'].apply(lambda x: x if x != 2400 else 0)

# Extract scheduled and actual departure hour
df['CRSDepHour'] = df['CRSDepTime'] // 100
df['DepHour'] = df['DepTime'] // 100
```

```python
def time_of_day(hour):
    if pd.isnull(hour):
        return 'Unknown'
    elif 5 <= hour < 12:
        return 'Morning'
    elif 12 <= hour < 17:
        return 'Afternoon'
    elif 17 <= hour < 21:
        return 'Evening'
    elif 21 <= hour or hour < 5:
        return 'Night'
    else:
        return 'Unknown'

df['Dep_TimeOfDay'] = df['DepHour'].apply(time_of_day)
df['CRSDep_TimeOfDay'] = df['CRSDepHour'].apply(time_of_day)
```

```python
import calendar
df['Day_Name'] = df['DayOfWeek'].apply(lambda x: calendar.day_name[x - 1])
```

```python
import numpy as np

def count_outliers(group):
    q1 = group.quantile(0.25)
    q3 = group.quantile(0.75)
    iqr = q3 - q1
    lower = q1 - 1.5 * iqr
    upper = q3 + 1.5 * iqr
    return ((group < lower) | (group > upper)).sum()

outlier_counts = df.groupby('Day_Name')['ArrDelay'].apply(count_outliers)
print(outlier_counts)
```

```python
# Function to calculate outliers for each airline
def count_outliers_by_airline(df):
    outlier_counts = {}

    for carrier in df['UniqueCarrier'].unique():
        group = df[df['UniqueCarrier'] == carrier]['ArrDelay'].dropna()
        Q1 = group.quantile(0.25)
        Q3 = group.quantile(0.75)
        IQR = Q3 - Q1
        lower_bound = Q1 - 1.5 * IQR
        upper_bound = Q3 + 1.5 * IQR
        outliers = group[(group < lower_bound) | (group > upper_bound)]
        outlier_counts[carrier] = len(outliers)

    return pd.Series(outlier_counts).sort_values(ascending=False)

# Call the function and display the result
outlier_counts = count_outliers_by_airline(df)
print(outlier_counts)
```

```python
import pandas as pd

# Step 1: Group by airline
grouped = df.groupby('UniqueCarrier')

# Step 2: Create empty list to store results
results = []

# Step 3: Loop through each airline
for airline, group in grouped:
    total_flights = len(group)
    median_delay = group['ArrDelay'].median()

    # IQR method applied within airline group
    Q1 = group['ArrDelay'].quantile(0.25)
    Q3 = group['ArrDelay'].quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = group[(group['ArrDelay'] < lower) | (group['ArrDelay'] > upper)]
    outlier_count = len(outliers)
    outlier_pct = (outlier_count / total_flights) * 100

    # Append to result list
    results.append({
        'UniqueCarrier': airline,
        'Total_Flights': total_flights,
        'Outliers': outlier_count,
        'Outlier_%': outlier_pct,
        'Median_ArrDelay': median_delay
    })

# Step 4: Create DataFrame
summary = pd.DataFrame(results).sort_values(by='Outliers', ascending=False)

# Step 5: Display result
print(summary)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# Compute correlation matrix
corr_matrix = df.select_dtypes(include='number').corr()

# Create a mask to hide the upper triangle and diagonal
mask = np.triu(np.ones_like(corr_matrix, dtype=bool))

# Plot heatmap without self-correlations
plt.figure(figsize=(16, 12))
sns.heatmap(corr_matrix,
            mask=mask,
            annot=True,
            fmt=".2f",
            cmap='coolwarm',
            square=True,
            cbar_kws={"shrink": .8},
            annot_kws={"size": 10})

# Rotate axis labels
plt.xticks(rotation=45, ha='right', fontsize=10)
plt.yticks(rotation=0, fontsize=10)

plt.title("Correlation Matrix (No Self-Correlation)", fontsize=16)
plt.tight_layout()
plt.show()
```
