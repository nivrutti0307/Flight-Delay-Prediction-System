```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.feature_selection import RFE
from sklearn.ensemble import RandomForestClassifier
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.tree import DecisionTreeClassifier
from xgboost import XGBClassifier
from lightgbm import LGBMClassifier
from sklearn.metrics import f1_score, roc_auc_score
import joblib
```

```python
# 2. Handle Missing Values
def handle_missing_data(df):
    # Fill delay-related columns with 0
    delay_cols = ['DELAY_DUE_CARRIER', 'DELAY_DUE_WEATHER', 'DELAY_DUE_NAS',
                 'DELAY_DUE_SECURITY', 'DELAY_DUE_LATE_AIRCRAFT']
    for col in delay_cols:
        df[col] = df[col].fillna(0)

    # Time-based columns: Impute with route-specific medians
    time_cols = ['DEP_TIME', 'DEP_DELAY', 'ARR_TIME', 'ARR_DELAY',
                'ELAPSED_TIME', 'AIR_TIME']

    for col in time_cols:
        # Calculate route medians
        route_medians = df.groupby(['ORIGIN', 'DEST'])[col].transform('median')
        # Fill missing values with route medians
        df[col] = df[col].fillna(route_medians)
        # Fill any remaining with overall median
        df[col] = df[col].fillna(df[col].median())

    return df

stratified_df = handle_missing_data(stratified_df.copy())
```

```python
# 3. Feature Engineering
def create_features(df):
    # Convert to datetime
    df['FL_DATE'] = pd.to_datetime(df['FL_DATE'])

    # Temporal features
    df['MONTH'] = df['FL_DATE'].dt.month
    df['DAY_OF_WEEK'] = df['FL_DATE'].dt.dayofweek  # Monday=0

    # Season mapping
    season_map = {
        12: 'Winter', 1: 'Winter', 2: 'Winter',
        3: 'Spring', 4: 'Spring', 5: 'Spring',
        6: 'Summer', 7: 'Summer', 8: 'Summer',
        9: 'Fall', 10: 'Fall', 11: 'Fall'
    }
    df['SEASON'] = df['MONTH'].map(season_map)

    # Time categories
    def get_time_category(time):
        if 0 <= time < 600:
            return 'Early_Morning'
        elif 600 <= time < 1200:
            return 'Morning'
        elif 1200 <= time < 1800:
            return 'Afternoon'
        else:
            return 'Evening'

    df['TIME_CATEGORY'] = df['CRS_DEP_TIME'].apply(get_time_category)

    return df

stratified_df = create_features(stratified_df)
```

```python
# 4. Define Features and Target
# Remove leaky and irrelevant features
leaky_features = ['DEP_TIME', 'DEP_DELAY', 'ARR_TIME', 'ARR_DELAY',
                 'ELAPSED_TIME', 'AIR_TIME', 'DELAY_DUE_CARRIER',
                 'DELAY_DUE_WEATHER', 'DELAY_DUE_NAS', 'DELAY_DUE_SECURITY',
                 'DELAY_DUE_LATE_AIRCRAFT', 'CANCELLATION_CODE', 'FlightDate']

unused_features = ['FL_NUMBER', 'DOT_CODE', 'AIRLINE_DOT', 'AIRLINE_CODE',
                  'EventId', 'DIVERTED', 'WHEELS_OFF', 'WHEELS_ON',
                  'TAXI_OUT', 'TAXI_IN', 'CANCELLED', 'FL_DATE',
                  'ORIGIN_CITY', 'DEST_CITY', 'AIRLINE']

# Features to keep
features = ['ORIGIN', 'DEST', 'CRS_DEP_TIME', 'CRS_ARR_TIME',
           'CRS_ELAPSED_TIME', 'DISTANCE', 'Type', 'Severity',
           'Precipitation(in)', 'MONTH', 'DAY_OF_WEEK', 'SEASON', 'TIME_CATEGORY']

# Create feature matrix and target
X = stratified_df[features]
y = stratified_df['DELAY_DUE_WEATHER_YN'].map({'Yes': 1, 'No': 0})  # Convert to binary
```

```python
# 5. Preprocessing Pipeline
# Categorical features
cat_features = ['ORIGIN', 'DEST', 'Type', 'Severity', 'SEASON', 'TIME_CATEGORY']
num_features = ['CRS_DEP_TIME', 'CRS_ARR_TIME', 'CRS_ELAPSED_TIME',
               'DISTANCE', 'Precipitation(in)', 'MONTH', 'DAY_OF_WEEK']

preprocessor = ColumnTransformer(
    transformers=[
        ('num', StandardScaler(), num_features),
        ('cat', OneHotEncoder(handle_unknown='ignore', sparse_output=False), cat_features)
    ])
```

```python
# 6. Train/Test Split
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, stratify=y, random_state=42
)

# 7. Feature Selection with RFE
# Preprocess training data
X_train_preprocessed = preprocessor.fit_transform(X_train)

# Get feature names
cat_encoder = preprocessor.named_transformers_['cat']
cat_feature_names = cat_encoder.get_feature_names_out(cat_features)
feature_names = np.concatenate([num_features, cat_feature_names])

# Initialize RFE
rfe_selector = RFE(
    estimator=RandomForestClassifier(n_estimators=100, random_state=42),
    n_features_to_select=8,  # Select 8 features
    step=0.1
)
rfe_selector.fit(X_train_preprocessed, y_train)

# Get selected feature names
selected_features = feature_names[rfe_selector.support_]
print("Selected Features:", selected_features)
```

## 'Logistic Regression', 'SVM', 'Decision Tree', 'Random Forest'

```python
from tqdm import tqdm
import time

import warnings
warnings.filterwarnings("ignore", category=UserWarning)
warnings.filterwarnings("ignore", category=FutureWarning)

# 8. Model Training
models = {
    'Logistic Regression': LogisticRegression(max_iter=1000, class_weight='balanced', random_state=42),
    'SVM': SVC(probability=True, class_weight='balanced', random_state=42),
    'Decision Tree': DecisionTreeClassifier(max_depth=8, random_state=42),
    'Random Forest': RandomForestClassifier(n_estimators=200, class_weight='balanced', random_state=42),
    'XGBoost': XGBClassifier(
        scale_pos_weight=sum(y==0)/sum(y==1),
        eval_metric='logloss',
        verbosity=0
    ),
    'LightGBM': LGBMClassifier(
        class_weight='balanced',
        random_state=42,
        verbose=-1
    )
}

# Create full pipeline
results = {}
best_model = None
best_f1 = 0

# Create tqdm progress bar
model_names = list(models.keys())
pbar = tqdm(model_names, desc="Training Models", unit="model")

for name in pbar:
    model = models[name]
    pbar.set_description(f"Training {name}")

    # Create pipeline
    pipeline = Pipeline([
        ('preprocessor', preprocessor),
        ('selector', rfe_selector),
        ('classifier', model)
    ])

    # Train model with timing
    start_time = time.time()
    pipeline.fit(X_train, y_train)
    training_time = time.time() - start_time

    # Evaluate
    y_pred = pipeline.predict(X_test)
    y_proba = pipeline.predict_proba(X_test)[:, 1]

    f1 = f1_score(y_test, y_pred)
    auc = roc_auc_score(y_test, y_proba)

    results[name] = {'F1': f1, 'AUC': auc, 'Time': training_time}

    # Track best model
    if f1 > best_f1:
        best_f1 = f1
        best_model = pipeline

    # Update progress description
    pbar.set_postfix_str(f"F1: {f1:.4f}, Time: {training_time:.1f}s")

# Close progress bar
pbar.close()

# Print results
print("\nModel Performance Summary:")
print("{:<20} {:<10} {:<10} {:<10}".format('Model', 'F1 Score', 'AUC', 'Time (s)'))
for name, metrics in results.items():
    print("{:<20} {:<10.4f} {:<10.4f} {:<10.1f}".format(
        name, metrics['F1'], metrics['AUC'], metrics['Time']
    ))

print(f"\nBest Model: {type(best_model.named_steps['classifier']).__name__}")
print(f"Best F1 Score: {best_f1:.4f}")
```

## 2. XGBClassifier

```python
import warnings
warnings.filterwarnings("ignore")

import pandas as pd
import numpy as np
import time
import joblib
from datetime import datetime

from sklearn.model_selection import train_test_split
from sklearn.feature_selection import RFE
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import f1_score, roc_auc_score, accuracy_score

from xgboost import XGBClassifier

import category_encoders as ce

# 1. Define Features and Target
features = ['ORIGIN', 'DEST', 'CRS_DEP_TIME', 'CRS_ARR_TIME',
            'CRS_ELAPSED_TIME', 'DISTANCE', 'Type', 'Severity',
            'Precipitation(in)', 'MONTH', 'DAY_OF_WEEK', 'SEASON', 'TIME_CATEGORY']

X = stratified_df[features]
y = stratified_df['DELAY_DUE_WEATHER_YN'].map({'Yes': 1, 'No': 0})

# 2. Target Encoding for Categorical Columns
categorical_cols = ['ORIGIN', 'DEST', 'Type', 'Severity', 'SEASON', 'TIME_CATEGORY']
encoder = ce.TargetEncoder(cols=categorical_cols)
X_encoded = encoder.fit_transform(X, y)

# 3. Train/Test Split
X_train, X_test, y_train, y_test = train_test_split(
    X_encoded, y, test_size=0.2, stratify=y, random_state=42
)

# 4. Feature Selection with RFE
rfe_selector = RFE(
    estimator=RandomForestClassifier(n_estimators=100, random_state=42, n_jobs=-1),
    n_features_to_select=12,
    step=0.1
)
rfe_selector.fit(X_train, y_train)

# Select Features
selected_features = X_train.columns[rfe_selector.support_]
X_train_selected = X_train[selected_features]
X_test_selected = X_test[selected_features]

print("Selected Features:", list(selected_features))

# 5. Train XGBoost Classifier
model = XGBClassifier(
    random_state=42,
    eval_metric='logloss',
    use_label_encoder=False,
    learning_rate=0.1,
    max_depth=5,
    subsample=0.9,
    colsample_bytree=0.9,
    n_estimators=200  # Reduced to 200
)

start_time = time.time()
print("\nTraining XGBoost...")
model.fit(X_train_selected, y_train)
print(f"\nTraining completed in {(time.time() - start_time) / 60:.1f} minutes")

# 6. Evaluate Model
y_pred = model.predict(X_test_selected)
y_proba = model.predict_proba(X_test_selected)[:, 1]

f1 = f1_score(y_test, y_pred)
auc = roc_auc_score(y_test, y_proba)
accuracy = accuracy_score(y_test, y_pred)

print("\nModel Performance:")
print(f"F1 Score: {f1:.4f}")
print(f"AUC: {auc:.4f}")
print(f"Accuracy: {accuracy:.4f}")

# 7. Create Deployment Pipeline (FIXED)
class FlightDelayPredictor:
    def __init__(self, encoder, selector, model, selected_features):
        self.encoder = encoder
        self.selector = selector
        self.model = model
        self.selected_features = selected_features  # Store selected feature names

    def predict_proba(self, X):
        X_encoded = self.encoder.transform(X)
        X_selected = X_encoded[self.selected_features]  # Use stored feature names
        return self.model.predict_proba(X_selected)

# Save model with selected features
final_predictor = FlightDelayPredictor(encoder, rfe_selector, model, selected_features)
joblib.dump(final_predictor, 'flight_delay_predictor.pkl')

# 8. Prediction Function
def predict_flight_delay(origin, destination, date,
                         weather_type='Unknown', severity='Unknown', precipitation=0.0):
    flight_date = datetime.strptime(date, '%Y-%m-%d')
    month = flight_date.month
    day_of_week = flight_date.weekday()

    season_map = {
        12: 'Winter', 1: 'Winter', 2: 'Winter',
        3: 'Spring', 4: 'Spring', 5: 'Spring',
        6: 'Summer', 7: 'Summer', 8: 'Summer',
        9: 'Fall', 10: 'Fall', 11: 'Fall'
    }
    season = season_map.get(month, 'Winter')

    input_data = pd.DataFrame({
        'ORIGIN': [origin],
        'DEST': [destination],
        'CRS_DEP_TIME': [800],
        'CRS_ARR_TIME': [1000],
        'CRS_ELAPSED_TIME': [120],
        'DISTANCE': [300],
        'Type': [weather_type],
        'Severity': [severity],
        'Precipitation(in)': [precipitation],
        'MONTH': [month],
        'DAY_OF_WEEK': [day_of_week],
        'SEASON': [season],
        'TIME_CATEGORY': ['Morning']
    })

    model = joblib.load('flight_delay_predictor.pkl')
    proba = model.predict_proba(input_data)[0][1]
    prediction = 1 if proba >= 0.5 else 0

    return prediction, proba

# 9. Example Prediction
print("\nExample prediction:")
pred, prob = predict_flight_delay('ABR', 'MSP', '2023-12-15',
                                  weather_type='Snow', severity='Severe', precipitation=0.5)
print(f"Prediction: {'Delayed' if pred == 1 else 'Not Delayed'}")
print(f"Probability: {prob:.1%}")
```

## 3. RandomForestClassifier

```python
import warnings
warnings.filterwarnings("ignore")

import pandas as pd
import numpy as np
import time
import joblib
from datetime import datetime

from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.feature_selection import RFE
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import f1_score, roc_auc_score, accuracy_score, confusion_matrix, classification_report
import category_encoders as ce

# 1. Define Features and Target
features = ['ORIGIN', 'DEST', 'CRS_DEP_TIME', 'CRS_ARR_TIME',
            'CRS_ELAPSED_TIME', 'DISTANCE', 'Type', 'Severity',
            'Precipitation(in)', 'MONTH', 'DAY_OF_WEEK', 'SEASON', 'TIME_CATEGORY']

X = stratified_df[features]
y = stratified_df['DELAY_DUE_WEATHER_YN'].map({'Yes': 1, 'No': 0})

# 2. Target Encoding for Categorical Columns
categorical_cols = ['ORIGIN', 'DEST', 'Type', 'Severity', 'SEASON', 'TIME_CATEGORY']
encoder = ce.TargetEncoder(cols=categorical_cols)
X_encoded = encoder.fit_transform(X, y)

# 3. Train/Test Split
X_train, X_test, y_train, y_test = train_test_split(
    X_encoded, y, test_size=0.2, stratify=y, random_state=42
)

# 4. Feature Selection with RFE
rfe_selector = RFE(
    estimator=RandomForestClassifier(n_estimators=100, random_state=42, n_jobs=-1),
    n_features_to_select=12,
    step=0.1
)
rfe_selector.fit(X_train, y_train)

# Select Features
selected_features = X_train.columns[rfe_selector.support_]
X_train_selected = X_train[selected_features]
X_test_selected = X_test[selected_features]

print("Selected Features:", list(selected_features))

# 5. Hyperparameter Tuning for Random Forest
param_grid = {
    'n_estimators': [200, 300],
    'max_depth': [15, 20, None],
    'min_samples_split': [5, 10],
    'min_samples_leaf': [2, 4],
    'max_features': ['sqrt', 0.8],
    'class_weight': ['balanced', None]
}

rf = RandomForestClassifier(random_state=42, n_jobs=-1)
grid_search = GridSearchCV(
    estimator=rf,
    param_grid=param_grid,
    scoring='accuracy',
    cv=3,
    n_jobs=-1,
    verbose=1
)

print("\nStarting Grid Search for Random Forest...")
start_time = time.time()
grid_search.fit(X_train_selected, y_train)
print(f"Grid Search completed in {(time.time() - start_time)/60:.1f} minutes")

best_rf = grid_search.best_estimator_
print("\nBest Parameters:", grid_search.best_params_)

# 6. Evaluate Model
y_pred = best_rf.predict(X_test_selected)
y_proba = best_rf.predict_proba(X_test_selected)[:, 1]

f1 = f1_score(y_test, y_pred)
auc = roc_auc_score(y_test, y_proba)
accuracy = accuracy_score(y_test, y_pred)

print("\nModel Performance:")
print(f"F1 Score: {f1:.4f}")
print(f"AUC: {auc:.4f}")
print(f"Accuracy: {accuracy:.4f}")

# Additional evaluation metrics
print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred))
print("\nClassification Report:")
print(classification_report(y_test, y_pred))

# 7. Create Deployment Pipeline
class FlightDelayPredictor:
    def __init__(self, encoder, selector, model, selected_features):
        self.encoder = encoder
        self.selector = selector
        self.model = model
        self.selected_features = selected_features

    def predict_proba(self, X):
        X_encoded = self.encoder.transform(X)
        X_selected = X_encoded[self.selected_features]
        return self.model.predict_proba(X_selected)

# Save model with selected features
final_predictor = FlightDelayPredictor(encoder, rfe_selector, best_rf, selected_features)
joblib.dump(final_predictor, 'flight_delay_predictor_rf.pkl')

# 8. Prediction Function
def predict_flight_delay(origin, destination, date,
                         weather_type='Unknown', severity='Unknown', precipitation=0.0):
    flight_date = datetime.strptime(date, '%Y-%m-%d')
    month = flight_date.month
    day_of_week = flight_date.weekday()

    season_map = {
        12: 'Winter', 1: 'Winter', 2: 'Winter',
        3: 'Spring', 4: 'Spring', 5: 'Spring',
        6: 'Summer', 7: 'Summer', 8: 'Summer',
        9: 'Fall', 10: 'Fall', 11: 'Fall'
    }
    season = season_map.get(month, 'Winter')

    input_data = pd.DataFrame({
        'ORIGIN': [origin],
        'DEST': [destination],
        'CRS_DEP_TIME': [800],
        'CRS_ARR_TIME': [1000],
        'CRS_ELAPSED_TIME': [120],
        'DISTANCE': [300],
        'Type': [weather_type],
        'Severity': [severity],
        'Precipitation(in)': [precipitation],
        'MONTH': [month],
        'DAY_OF_WEEK': [day_of_week],
        'SEASON': [season],
        'TIME_CATEGORY': ['Morning']
    })

    model = joblib.load('flight_delay_predictor_rf.pkl')
    proba = model.predict_proba(input_data)[0][1]
    prediction = 1 if proba >= 0.5 else 0

    return prediction, proba

# 9. Example Prediction
print("\nExample prediction:")
pred, prob = predict_flight_delay('ABR', 'MSP', '2023-12-15',
                                  weather_type='Snow', severity='Severe', precipitation=0.2)
print(f"Prediction: {'Delayed' if pred == 1 else 'Not Delayed'}")
print(f"Probability: {prob:.1%}")
```
