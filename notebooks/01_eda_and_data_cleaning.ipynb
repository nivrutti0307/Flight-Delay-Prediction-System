```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

# Set up visual style
sns.set(style="whitegrid")
```

```python
import pandas as pd

# Adjust the path based on your folder structure
file_path = '/content/drive/MyDrive/split of the file/flight_2019_2023_data.csv'

# Load the CSV
df = pd.read_csv(file_path)

# Display basic info
df.head()
```

```python
# Check basic info
df.info()

# Check for missing values
print(df.isnull().sum())

# Check number of unique values per column
print(df.nunique())

# Data types of each column
print(df.dtypes)

# Summary statistics
df.describe(include='all')
```

```python
# % missing values
missing = df.isnull().mean().sort_values(ascending=False) * 100
missing[missing > 0]
```

```python
missing[missing > 0].plot(kind='barh', figsize=(10,6), title='Missing Value Percentage')
plt.xlabel("Percentage")
plt.show()
```

```python
# Basic delay stats
delay_cols = ['DEP_DELAY', 'ARR_DELAY', 'DELAY_DUE_CARRIER', 'DELAY_DUE_WEATHER', 'DELAY_DUE_NAS', 'DELAY_DUE_SECURITY', 'DELAY_DUE_LATE_AIRCRAFT']
df[delay_cols].describe()
```

```python
sns.histplot(df['ARR_DELAY'].dropna(), bins=100, kde=True)
plt.title('Arrival Delay Distribution')
plt.xlim(-100, 500)
plt.show()
```

```python
# Cancellations
print("Cancelled flights:", df['CANCELLED'].sum())
sns.countplot(x='CANCELLED', data=df)
plt.title("Cancelled vs Non-Cancelled Flights")
plt.xticks([0, 1], ['No', 'Yes'])
plt.show()

# Diversions
print("Diverted flights:", df['DIVERTED'].sum())
sns.countplot(x='DIVERTED', data=df)
plt.title("Diverted vs Non-Diverted Flights")
plt.xticks([0, 1], ['No', 'Yes'])
plt.show()
```

```python
# Top 10 Origin Cities
df['ORIGIN_CITY'].value_counts().head(10).plot(kind='bar', title='Top 10 Origin Cities')
plt.xticks(rotation=45)
plt.show()

# Top Airlines by flights
df['AIRLINE'].value_counts().head(10).plot(kind='bar', title='Top Airlines by Flight Count')
plt.xticks(rotation=45)
plt.show()
```

```python
df['FL_DATE'] = pd.to_datetime(df['FL_DATE'], dayfirst=True)
```

```python
# Monthly mean arrival delay
monthly_delay = df.groupby(df['FL_DATE'].dt.to_period('M'))['ARR_DELAY'].mean()

monthly_delay.plot(kind='line', figsize=(12,5), title="Monthly Average Arrival Delay")
plt.ylabel("Average Delay (min)")
plt.xlabel("Month")
plt.show()
```

```python
# Correlation heatmap
corr_cols = ['ARR_DELAY', 'DEP_DELAY', 'DELAY_DUE_WEATHER', 'Precipitation(in)']
sns.heatmap(df[corr_cols].corr(), annot=True, cmap='coolwarm')
plt.title("Correlation Heatmap")
plt.show()
```

```python
# Severity impact
sns.boxplot(data=df, x='Severity', y='ARR_DELAY')
plt.title("Arrival Delay by Weather Severity")
plt.show()

# Weather Type impact
sns.boxplot(data=df, x='Type', y='ARR_DELAY')
plt.title("Arrival Delay by Weather Type")
plt.xticks(rotation=45)
plt.show()
```
